# Reading From CSV and Write to Bronze Layer

## crm_cust_info

```python
df = (spark.read
      .option("header", "true")
      .option("inferSchema", "true")
      .csv("/Volumes/workspace/bronze/data_source/source_crm/cust_info.csv")
      )

df.write.mode("overwrite").saveAsTable("bronze.crm_cust_info")
```
