# Transformation Logic

```python
query = """
SELECT 
        *

FROM workspace.silver.crm_products pr
LEFT JOIN workspace.silver.erp_product_category pc
    ON pr.category_id = pc.category_id
"""
df = spark.sql(query)
df.display()
```

# Writing Gold Table

```python
df.write.mode("overwrite").format("delta").saveAsTable("gold.dim_customers")
```
